go: they need the GPUs idle.

Every Run All clones your repository afresh: push to the default branch and leave `COMMIT`
empty (or set it to the new commit). For the submission: make this notebook public, set

```python
REPO_URL = "https://github.com/tokha-t/decik.git"  # your repository
COMMIT = ""  # the commit you submit; empty = the default branch
```

```python
import os, subprocess
names = subprocess.run(["nvidia-smi", "--query-gpu=name", "--format=csv,noheader"],
                       capture_output=True, text=True, check=True).stdout
print(names)
assert "T4" in names, "Switch the accelerator to GPU T4 x2"
packages = ["soup-cli[train]==0.77.0", "transformers==5.16.1", "peft==0.20.0",
            "trl==0.29.1", "bitsandbytes==0.50.2", "accelerate==1.14.0", "pytest"]
subprocess.run(["python", "-m", "pip", "install", "-q", *packages], check=True)
```

```python
import base64, os, shutil, subprocess
token = ""
try:
    from kaggle_secrets import UserSecretsClient
    token = UserSecretsClient().get_secret("GITHUB_TOKEN")
except Exception as exc:
    print("No GITHUB_TOKEN secret, cloning without it:", type(exc).__name__)
auth = []
if token:  # the token goes in a header for this one command: never in a URL, a file or the output
    basic = base64.b64encode(f"x-access-token:{token}".encode()).decode()
    auth = ["-c", f"http.extraHeader=Authorization: Basic {basic}"]
shutil.rmtree("/kaggle/working/repo", ignore_errors=True)  # always a fresh copy of your code
clone = subprocess.run(["git", *auth, "clone", "-q", REPO_URL, "/kaggle/working/repo"],
                       capture_output=True, text=True)
if clone.returncode:
    raise SystemExit("git clone failed: " + clone.stderr)
if COMMIT:
    subprocess.run(["git", "-C", "/kaggle/working/repo", "checkout", "-q", COMMIT], check=True)
print(subprocess.run(["git", "-C", "/kaggle/working/repo", "log", "-1", "--format=%H %s"],
                     capture_output=True, text=True).stdout)
```

```python
import subprocess
REPO = "/kaggle/working/repo"
subprocess.run(["python", "task/prepare_data.py"], cwd=REPO, check=True)
subprocess.run(["python", "-m", "pytest", "solution", "-q"], cwd=REPO, check=True)
```

```python
subprocess.run(["python", "harness/run_task.py", "--repeats", "2"], cwd=REPO, check=True)
```

```python
import json, pathlib, shutil, subprocess
subprocess.run(["python", "harness/check_rules.py"], cwd=REPO, check=True)
subprocess.run(["python", "harness/score.py", "runs/result.json"], cwd=REPO, check=True)
report = pathlib.Path("/kaggle/working/validation")
report.mkdir(exist_ok=True)
for name in ("result.json", "checks.json"):
    shutil.copy2(pathlib.Path(REPO) / "runs" / name, report / name)
commit = subprocess.check_output(["git", "rev-parse", "HEAD"], cwd=REPO, text=True).strip()
(report / "manifest.json").write_text(json.dumps({"commit": commit, "repeats": 2,
    "solution_env": {k: v for k, v in os.environ.items() if k.startswith("SKC_")}}, indent=2))
print("Validation evidence saved in", report)
```
